```python
import pandas as pd
import numpy as np
from sklearn.metrics import accuracy_score, classification_report
import pickle
data = pd.read_csv(r'C:\Users\AdamCage\Documents\Света\тест_диплома\heart.csv')
with open('model_09f11181bfdc.pkl', 'rb') as f:
    model = pickle.load(f)

X = data.drop('HeartDisease', axis=1)
y = data['HeartDisease']

# 3️⃣ Предсказания
y_pred = model["pipeline"].predict(X)

# 4️⃣ Точность (только для классификации!)
acc = accuracy_score(y, y_pred)
print(f"✅ Accuracy: {acc:.4f} ({acc*100:.2f}%)")

# 📊 Дополнительно: подробный отчёт
print("\n📋 Classification Report:")
print(classification_report(y, y_pred))

# 💡 Сохранить предсказания обратно в датасет (опционально)
data['prediction'] = y_pred
```
